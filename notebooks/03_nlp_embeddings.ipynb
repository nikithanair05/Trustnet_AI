```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
profiles = [
    "Pay registration fee for job interview",
    "Official HR recruitment account",
    "Earn money instantly through crypto investment",
    "Photography and travel blogger",
    "Urgent investment opportunity guaranteed profit",
    "Personal fitness trainer and nutrition coach"
]
```

```python
embeddings = model.encode(profiles)
```

```python
embeddings.shape
```

Meaning:

6 sentences
384-dimensional meaning vectors


```python
similarity = cosine_similarity([embeddings[0]], embeddings)

print(similarity)
```

Compares:

"Pay registration fee for job interview"

against all profiles.

```python
similarity_scores = similarity[0]

results = pd.DataFrame({
    'Profile': profiles,
    'Similarity': similarity_scores
})

results = results.sort_values(
    by='Similarity',
    ascending=False
)

print(results)
```

```python
new_text = [
    "Send payment immediately to confirm interview slot"
]
```

```python
new_embedding = model.encode(new_text)
```

```python
new_similarity = cosine_similarity(new_embedding, embeddings)

print(new_similarity)
```
